## Exploratory Data Analysis (EDA)

### Sales by Category


```python
import matplotlib.pyplot as plt

category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(6,4))
category_sales.plot(kind="bar")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

### Profit by Category

```python
category_profit = df.groupby("Category")["Profit"].sum()

plt.figure(figsize=(6,4))
category_profit.plot(kind="bar")

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum()
print(category_profit)
```

### Sales by Region

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

print(region_sales)

plt.figure(figsize=(7,5))
region_sales.plot(kind="bar", color="skyblue")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
print(region_sales)
```

```python
region_profit = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)

print(region_profit)

plt.figure(figsize=(7,5))
region_profit.plot(kind="bar", color="lightgreen")

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()

monthly_sales.index = monthly_sales.index.astype(str)

print(monthly_sales.head())

plt.figure(figsize=(12,5))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.grid(True)

plt.show()
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

print(segment_sales)

plt.figure(figsize=(7,5))
segment_sales.plot(kind="bar", color="coral")

plt.title("Total Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
top_states = df.groupby("State")["Sales"].sum().sort_values(ascending=False).head(10)

print(top_states)

plt.figure(figsize=(10,6))
top_states.plot(kind="bar", color="mediumpurple")

plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df["Discount"], df["Profit"], alpha=0.5)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.grid(True)

plt.show()
```

# Business Insights

1. **Technology** generated the highest profit among all product categories, making it the most profitable segment for the business.

2. **Furniture** generated significantly lower profit despite good sales, indicating that pricing, discounts, or operational costs may be reducing profitability.

3. The **West** region recorded the highest sales and profit, making it the strongest-performing region.

4. The **Central** region earned the lowest profit, suggesting opportunities to improve pricing strategies and operational efficiency.

5. The **Consumer** segment contributed the highest sales, making it the company's largest customer group.

6. **California** generated the highest sales among all states, followed by **New York** and **Texas**.

7. The monthly sales trend shows fluctuations in sales over time, indicating seasonal demand patterns.

8. The Discount vs Profit analysis suggests that higher discounts are often associated with lower or negative profits. The company should carefully evaluate its discount strategy to maintain profitability.

# Conclusion

This project analyzed the Sample Superstore dataset to evaluate sales performance, profitability, customer segments, regional performance, and the impact of discounts on profit. The analysis revealed that Technology is the most profitable category, the West region generates the highest sales and profit, and the Consumer segment contributes the largest share of sales. Additionally, the analysis showed that higher discounts are often associated with lower profitability. The insights obtained from this project can help businesses make informed decisions regarding pricing, marketing strategies, inventory management, and regional business expansion.
